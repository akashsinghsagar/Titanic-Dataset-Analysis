## Data Visualization 📊

In this step, we will **visualize the Titanic dataset** to understand patterns and relationships between features.  

**Objectives of Visualization:**

1. **Visualize distributions**  
   - Understand how numeric features like `Age` and `Fare` are distributed.  
   - Identify outliers or unusual patterns.

2. **Visualize categorical counts**  
   - Check the frequency of categories such as `Survived`, `Pclass`, `Sex`, and `Embarked`.  

3. **Explore relationships between features**  
   - Analyze how survival depends on features like `Sex`, `Pclass`, and combinations of multiple features.  
   - Use barplots, countplots, and other visualizations to discover hidden patterns.

> **Goal:** Visualization helps in making informed decisions about data cleaning, feature engineering, and understanding the factors affecting survival.


````python
# Count of survivors
sns.countplot(x='survived', data=data, palette='pastel')
plt.title('Survival Count')
plt.show()

# Count by passenger class
sns.countplot(x='pclass', data=data, palette='coolwarm')
plt.title('Passenger Class Count')
plt.show()
````

````python
sns.barplot(x='sex', y='survived', data=data, palette='muted')
plt.title('Survival Rate by Gender')
plt.show()
````

Higher bar = higher survival rate

Compare male vs female survival rates

FEMALE survived comparatively higher then MENS

````python
sns.barplot(x='pclass', y='survived', data=data, palette='Set2')
plt.title('Survival Rate by Passenger Class')
plt.show()
````

````python
plt.figure(figsize=(8,4))
sns.histplot(data=data, x='sibsp', hue='survived', multiple='stack', palette='pastel', bins=8)
plt.title('Survival Count by Number of Siblings/Spouses (SibSp)')
plt.xlabel('SibSp')
plt.ylabel('Count')
plt.show()

# Parch vs Survived
plt.figure(figsize=(8,4))
sns.histplot(data=data, x='parch', hue='survived', multiple='stack', palette='muted', bins=8)
plt.title('Survival Count by Number of Parents/Children (Parch)')
plt.xlabel('Parch')
plt.ylabel('Count')
plt.show()
````

hue='Survived' separates the bars for survivors (1) and non-survivors (0).

multiple='stack' stacks survivors and non-survivors for comparison.

Helps to see how family size affects survival.
